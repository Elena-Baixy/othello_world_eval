```python
# Function to extract cell content and outputs
def get_cell_info(cell):
    cell_type = cell['cell_type']
    source = ''.join(cell['source'])
    outputs = []
    if 'outputs' in cell:
        for output in cell['outputs']:
            if output.get('output_type') == 'stream':
                outputs.append(''.join(output.get('text', [])))
            elif output.get('output_type') == 'execute_result':
                data = output.get('data', {})
                if 'text/plain' in data:
                    outputs.append(''.join(data['text/plain']))
            elif output.get('output_type') == 'error':
                outputs.append(f"ERROR: {output.get('ename', '')}: {output.get('evalue', '')}")
    return cell_type, source, outputs

# Show all markdown cells to understand the structure
print("=== MARKDOWN CELLS ===")
for i, cell in enumerate(notebook['cells']):
    if cell['cell_type'] == 'markdown':
        source = ''.join(cell['source'])
        print(f"\n--- Cell {i} ---")
        print(source[:500])
```
